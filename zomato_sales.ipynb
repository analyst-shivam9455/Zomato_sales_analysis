# Zomato Sales Analysis

```python
import pandas as pd
```

### loading Data

```python
df = pd.read_csv('zomato.csv')
pd.set_option('display.max_columns', 20)
```

```python
# Now we check the shape of the dataset
df.shape 
```

```python
# now we take overviwe of the dataset by seeing few rows

df.head()
```

```python
# drop unnecessary columns

df.drop(columns=['url','address','phone','dish_liked','reviews_list'],inplace=True)
```

```python
df.head()
```

```python
# see the information about the dataset

df.info()
```

```python
# check for missing(null) values

df.isnull().sum()
```

```python
# correct rate column by split rating and convert it into right data type

df['rate'] = df['rate'].str.split('/').str.get(0)  
df['rate'] = pd.to_numeric(df['rate'], errors='coerce') 
```

```python
# filling null values with mean value

df['rate'] = df['rate'].fillna(df['rate'].mean())
```

```python
# converting datatype and filling null values with mean value

df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce') 
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].fillna(df['approx_cost(for two people)'].mean())
```

```python
# see the ststatistics of the dataset
df.describe()
```

```python
# saving cleaned data for further visualization

df.to_csv('zomato_cleaned_data.csv',index=False)
```
